```python
# Install the official Perplexity Python client
!pip install perplexityai

# Import and setup
import os
from dotenv import load_dotenv
from perplexity import Perplexity

# Load environment variables from .env file
load_dotenv()

# Initialize client
client = Perplexity()

# Example: Chat completion (use a Claude-accessible model if available via Pro, e.g., 'sonar-pro' or check docs for latest)
response = client.chat.completions.create(
    model="sonar-pro",  # Pro model with advanced capabilities; alternatives: 'pplx-70b-online', check https://docs.perplexity.ai for Claude options [web:8][web:20]
    messages=[
        {"role": "system", "content": "You are Claude Opus, a helpful AI assistant."},
        {"role": "user", "content": "Explain quantum computing in simple terms."}
    ],
    max_tokens=1000,
    temperature=0.7
)

print(response.choices[0].message.content)
```
